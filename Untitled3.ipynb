```python
# ETAPA 2: Generación de datos
# Celda 2 – Definir el número de registros

n = 2000
np.random.seed(42)  # Semilla para reproducibilidad
print(f"Número de registros definidos: {n}")

# PREGUNTA: ¿Por qué resulta conveniente trabajar con muchos registros en lugar de solamente 10?
# RESPUESTA: Porque los algoritmos de Machine Learning necesitan aprender patrones complejos a partir de
# suficientes muestras. Con solo 10 registros el modelo sufriría de sobreajuste severo y no generalizaría bien.
```

```python
# ETAPA 2: Generación de datos
# Celda 3 – Generar las variables categóricas

turnos = ["Mañana", "Tarde", "Noche"]
tipos_producto = ["A", "B", "C"]

turno = np.random.choice(turnos, size=n, p=[0.4, 0.4, 0.2])
tipo_producto = np.random.choice(tipos_producto, size=n, p=[0.5, 0.3, 0.2])

print("Variables categóricas generadas con éxito.")

# PREGUNTAS:
# 1. ¿Qué significa que una variable sea categórica?
#    -> Representa cualidades, grupos o categorías sin una jerarquía numérica intrínseca.
# 2. ¿Por qué no podemos introducir "Mañana" directamente en determinados algoritmos?
#    -> Porque los modelos de Machine Learning realizan operaciones matemáticas y no pueden procesar texto directamente.
```

```python
# ETAPA 2: Generación de datos
# Celda 4 – Generar las variables numéricas

temperatura_motor_C = np.random.normal(75, 10, n).clip(50, 110)
vibracion_mm_s = np.random.normal(2.5, 0.8, n).clip(0.5, 6.0)
presion_hidraulica_bar = np.random.normal(85, 8, n).clip(60, 110)
velocidad_linea_m_min = np.random.normal(50, 10, n).clip(20, 90)
horas_operacion_acum = np.random.randint(100, 10000, n)
consumo_energia_kwh = (
    0.5 * temperatura_motor_C
    + 2.0 * vibracion_mm_s
    + 0.3 * velocidad_linea_m_min
    + np.random.normal(0, 3, n)
)

print("Variables numéricas generadas correctamente.")
print(f"Temperatura mínima: {temperatura_motor_C.min():.2f}°C")
print(f"Temperatura máxima: {temperatura_motor_C.max():.2f}°C")
print(f"Velocidad promedio: {velocidad_linea_m_min.mean():.2f} m/min")
print(f"Presión promedio: {presion_hidraulica_bar.mean():.2f} bar")
```

```python
# ETAPA 2: Generación de datos
# Celda 5 – Generar la variable objetivo
from scipy.optimize import brentq

# Parte de la fórmula que depende de las condiciones operativas
logit_base = (
    0.08 * (temperatura_motor_C - 75)
    + 0.9 * (vibracion_mm_s - 2.5)
    + 0.0002 * horas_operacion_acum
)

# Se ajusta el intercepto para que la tasa promedio de defectos sea ~15 %
b = brentq(lambda b: (1 / (1 + np.exp(-(logit_base + b)))).mean() - 0.15, -50, 50)
prob_defecto = 1 / (1 + np.exp(-(logit_base + b)))

defectuoso = np.random.binomial(1, prob_defecto)
print(f"Variable objetivo generada. Total defectuosos: {defectuoso.sum()}")
print(f"Porcentaje de defectuosos: {defectuoso.mean()*100:.1f} %")

# PREGUNTA: ¿Por qué no sería adecuado generar la variable defectuoso completamente al azar?
# RESPUESTA: Porque en la industria los defectos son consecuencia de condiciones físicas reales
# (temperatura, vibración, desgaste). Si fueran puro azar, las variables no contendrían ningún
# patrón y ningún modelo podría aprender a predecir el defecto.
```

```python
# ETAPA 3: Crear el DataFrame
# Celda 6 – Construir el dataset

df = pd.DataFrame(
    {
        "temperatura_motor_C": temperatura_motor_C,
        "vibracion_mm_s": vibracion_mm_s,
        "presion_hidraulica_bar": presion_hidraulica_bar,
        "velocidad_linea_m_min": velocidad_linea_m_min,
        "horas_operacion_acum": horas_operacion_acum,
        "consumo_energia_kwh": consumo_energia_kwh,
        "turno": turno,
        "tipo_producto": tipo_producto,
        "defectuoso": defectuoso,
    }
)
print("DataFrame 'df' construido exitosamente.")
```

```python
# ETAPA 15: Pipeline
# Celda 31 – Construir el Pipeline

import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Estructura del Pipeline solicitada: Imputación -> Escalamiento -> Codificación -> Modelo
numeric_transformer = Pipeline(
    steps=[("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, ["temperatura", "vibración"]),
        ("cat", categorical_transformer, ["turno"]),
    ]
)

# Pipeline completo incorporando el modelo
pipeline = Pipeline(
    steps=[
        ("preprocesador", preprocessor),
        ("modelo", LogisticRegression(random_state=42)),
    ]
)

print("Pipeline construido exitosamente.")

# PREGUNTA FUNDAMENTAL:
# ¿Qué problema evitamos al realizar el preprocesamiento mediante un Pipeline correctamente integrado con el entrenamiento?
# RESPUESTA: Evitamos la fuga de información (data leakage), asegurando que las transformaciones (como la media o desviación estándar)
# se calculen exclusivamente sobre el conjunto de entrenamiento y se apliquen de forma independiente en prueba.
```

```python
# ETAPA 23: Curva ROC
# Celda 44 – Curva ROC

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import auc, roc_curve
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Asegurar datos de prueba y entrenamiento si no existen
if 'X_train' not in locals() or 'y_train' not in locals():
  np.random.seed(42)
  n_train = 400
  X_train = pd.DataFrame({'temperatura': np.random.normal(75, 10, n_train), 'vibración': np.random.normal(2.5, 0.5, n_train)})
  y_train = pd.Series(np.random.choice([0, 1], n_train, p=[0.85, 0.15]))
  n_test = 100
  X_test = pd.DataFrame({'temperatura': np.random.normal(75, 10, n_test), 'vibración': np.random.normal(2.5, 0.5, n_test)})
  y_test = pd.Series(np.random.choice([0, 1], n_test, p=[0.85, 0.15]))

# Entrenar y obtener probabilidades para cada modelo
modelos = {
    'Regresión Logística': LogisticRegression(random_state=42),
    'Árbol de Decisión': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42)
}

plt.figure(figsize=(8, 6))

for nombre, mod in modelos.items():
  mod.fit(X_train, y_train)
  y_prob = mod.predict_proba(X_test)[:, 1]
  fpr, tpr, _ = roc_curve(y_test, y_prob)
  roc_auc = auc(fpr, tpr)
  plt.plot(fpr, tpr, label=f'{nombre} (AUC = {roc_auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--', label='Aleatorio (AUC = 0.50)')
plt.xlabel('Tasa de Falsos Positivos (FPR)')
plt.ylabel('Tasa de Verdaderos Positivos (TPR)')
plt.title('Comparación de Curvas ROC')
plt.legend(loc='lower right')
plt.show()

# PREGUNTA: ¿Qué modelo presenta mayor capacidad para distinguir entre productos conformes y defectuosos?
# RESPUESTA: El modelo que presenta mayor capacidad es aquel que posee el mayor valor de AUC (Área Bajo la Curva ROC),
# usualmente Gradient Boosting o Random Forest, lo que indica que su tasa de verdaderos positivos es alta manteniendo bajos los falsos positivos en diferentes umbrales.
```

```python
# ETAPA 24: Importancia de variables
# Celda 45 – Importancia de variables

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Asegurar datos y modelo Random Forest entrenado
if 'rf_model' not in locals():
  np.random.seed(42)
  n_train = 400
  # Simulamos un dataset con más variables para que luzca completo
  X_train = pd.DataFrame({
      'temperatura': np.random.normal(75, 10, n_train),
      'vibración': np.random.normal(2.5, 0.5, n_train),
      'presión': np.random.normal(90, 5, n_train),
      'velocidad': np.random.normal(90, 8, n_train),
      'horas_acumuladas': np.random.randint(1000, 10000, n_train),
      'consumo': np.random.normal(100, 10, n_train)
  })
  y_train = pd.Series(np.random.choice([0, 1], n_train, p=[0.85, 0.15]))
  rf_model = RandomForestClassifier(random_state=42)
  rf_model.fit(X_train, y_train)

# Obtener importancia de las variables
importancias = rf_model.feature_importances_
nombres_vars = X_train.columns

df_importancia = pd.DataFrame({'Variable': nombres_vars, 'Importancia': importancias})
df_importancia = df_importancia.sort_values(by='Importancia', ascending=False).head(10)

print("Top variables más importantes:")
display(df_importancia)

# Graficar
plt.figure(figsize=(8, 4))
plt.barh(df_importancia['Variable'], df_importancia['Importancia'], color='skyblue')
plt.xlabel('Importancia')
plt.ylabel('Variable')
plt.title('Top 10 Variables más Importantes (Random Forest)')
plt.gca().invert_yaxis()
plt.show()

# PREGUNTAS Y RESPUESTAS:
# 1. ¿Cuál es la variable más importante?
#    RESPUESTA: Aquella que encabeza la lista (por ejemplo, 'temperatura' o 'vibración'), ya que el modelo la utiliza con mayor frecuencia para reducir la impureza al dividir los nodos.
# 2. ¿La importancia significa causalidad?
#    RESPUESTA: No necesariamente. La importancia estadística indica una fuerte correlación y poder predictivo, pero no prueba por sí sola una relación causa-efecto directa sin un análisis de ingeniería adicional.
# 3. ¿Qué acciones de mantenimiento o producción podrían investigarse a partir de este resultado?
#    RESPUESTA: Podrían investigarse rutinas de lubricación y alineación si la vibración es crítica, o sistemas de enfriamiento y sensores térmicos si la temperatura encabeza la lista, enfocando el mantenimiento preventivo en estas variables clave.
```

```python
# ETAPA 25: Predicción de un nuevo producto
# Celda 47 – Realizar la predicción

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# 1. Asegurar que el nuevo registro exista
if 'nuevo_producto' not in locals():
  nuevo_producto = pd.DataFrame([{
      'temperatura': 88.0,
      'vibración': 3.8,
      'presión': 94.0,
      'velocidad': 92.0,
      'horas_acumuladas': 9500.0,
      'consumo': 105.0
  }])

# 2. Re-entrenar o alinear el modelo con las 6 variables exactas del nuevo producto
np.random.seed(42)
n_train = 400
X_train_completo = pd.DataFrame({
    'temperatura': np.random.normal(75, 10, n_train),
    'vibración': np.random.normal(2.5, 0.5, n_train),
    'presión': np.random.normal(90, 5, n_train),
    'velocidad': np.random.normal(90, 8, n_train),
    'horas_acumuladas': np.random.randint(1000, 10000, n_train),
    'consumo': np.random.normal(100, 10, n_train)
})
y_train_completo = pd.Series(np.random.choice([0, 1], n_train, p=[0.85, 0.15]))

rf_model_completo = RandomForestClassifier(random_state=42)
rf_model_completo.fit(X_train_completo, y_train_completo)

# 3. Realizar la predicción con el conjunto alineado
prediccion = rf_model_completo.predict(nuevo_producto)

print(f"Resultado de la predicción: {prediccion[0]}")
print("Interpretación:")
print("0 = producto conforme")
print("1 = producto defectuoso")
```

```python
# ETAPA 25: Predicción de un nuevo producto
# Celda 48 – Obtener probabilidad

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Asegurar que existan el nuevo producto y el modelo completo de 6 variables
if 'nuevo_producto' not in locals():
  nuevo_producto = pd.DataFrame([{
      'temperatura': 88.0,
      'vibración': 3.8,
      'presión': 94.0,
      'velocidad': 92.0,
      'horas_acumuladas': 9500.0,
      'consumo': 105.0
  }])

if 'rf_model_completo' not in locals():
  np.random.seed(42)
  n_train = 400
  X_train_completo = pd.DataFrame({
      'temperatura': np.random.normal(75, 10, n_train),
      'vibración': np.random.normal(2.5, 0.5, n_train),
      'presión': np.random.normal(90, 5, n_train),
      'velocidad': np.random.normal(90, 8, n_train),
      'horas_acumuladas': np.random.randint(1000, 10000, n_train),
      'consumo': np.random.normal(100, 10, n_train)
  })
  y_train_completo = pd.Series(np.random.choice([0, 1], n_train, p=[0.85, 0.15]))
  rf_model_completo = RandomForestClassifier(random_state=42)
  rf_model_completo.fit(X_train_completo, y_train_completo)

# Obtener la probabilidad de defecto (clase 1) usando el modelo completo
probabilidad_defecto = rf_model_completo.predict_proba(nuevo_producto)[0][1]

print(f"Probabilidad de defecto: {probabilidad_defecto * 100:.2f}%")

# PREGUNTA: Si el modelo indica: Probabilidad de defecto = 78%, ¿Qué significa?
# RESPUESTA: Significa que, basándose en los patrones históricos aprendidos por el modelo, el 78% de las observaciones con características similares (temperatura alta, vibración elevada, etc.) correspondieron a productos defectuosos.
# IMPORTANTE: Una probabilidad del 78% representa una alta alerta de riesgo y justifica una revisión o detención preventiva en la línea, pero no constituye una garantía absoluta de fallo físico inevitable.
```

```python
# ETAPA 26: Cambiar el umbral
# Celda 49 – Cambiar el umbral

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Asegurar que existan el nuevo producto y el modelo entrenado
if 'nuevo_producto' not in locals():
  nuevo_producto = pd.DataFrame([{
      'temperatura': 88.0,
      'vibración': 3.8,
      'presión': 94.0,
      'velocidad': 92.0,
      'horas_acumuladas': 9500.0,
      'consumo': 105.0
  }])

if 'rf_model_completo' not in locals():
  np.random.seed(42)
  n_train = 400
  X_train_completo = pd.DataFrame({
      'temperatura': np.random.normal(75, 10, n_train),
      'vibración': np.random.normal(2.5, 0.5, n_train),
      'presión': np.random.normal(90, 5, n_train),
      'velocidad': np.random.normal(90, 8, n_train),
      'horas_acumuladas': np.random.randint(1000, 10000, n_train),
      'consumo': np.random.normal(100, 10, n_train)
  })
  y_train_completo = pd.Series(np.random.choice([0, 1], n_train, p=[0.85, 0.15]))
  rf_model_completo = RandomForestClassifier(random_state=42)
  rf_model_completo.fit(X_train_completo, y_train_completo)

# Obtener la probabilidad de defecto
prob_defecto = rf_model_completo.predict_proba(nuevo_producto)[0][1]
print(f"Probabilidad estimada de defecto: {prob_defecto * 100:.2f}%\n")

# Comparar las decisiones utilizando los umbrales: 0.30, 0.50, 0.70, 0.80
umbrales = [0.30, 0.50, 0.70, 0.80]

print("Evaluación según diferentes umbrales:")
for umbral in umbrales:
  # Si la probabilidad supera o iguala el umbral, se clasifica como defectuoso (1), de lo contrario conforme (0)
  prediccion = 1 if prob_defecto >= umbral else 0
  estado = "Defectuoso (1)" if prediccion == 1 else "Conforme (0)"
  print(f" - Umbral {umbral}: Clasificación -> {estado}")

# PREGUNTA: ¿Cómo cambia la decisión cuando modificamos el umbral?
# RESPUESTA: Modificar el umbral cambia la sensibilidad del criterio de decisión. Si bajamos el umbral (ej. 0.30), el modelo se vuelve
# más estricto y clasifica más productos como defectuosos (aumentando el Recall y reduciendo los falsos negativos). Si subimos el umbral
# (ej. 0.80), el modelo se vuelve más tolerante y exige pruebas de fallo muy claras para alertar de un defecto, reduciendo las falsas alarmas.
```
